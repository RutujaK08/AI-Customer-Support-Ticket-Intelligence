```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")
```

```python
# Load the dataset

df = pd.read_csv("../Dataset/raw/customer_support_tickets.csv"
)
print("Dataset sucessfully loaded")
```

```python
print("Number of Rows:", df.shape[0])

print("Number of Columns:", df.shape[1])
```

```python
# Data quality issues
# Check missing values in each column

missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Calculate missing percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df)
    ) * 100
})

missing_report = missing_report.sort_values(
    by="Missing_Percentage",
    ascending=False
)

missing_report
```

```python
# Check duplicate rows

duplicate_rows = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_rows)
```

```python
# Check duplicate ticket IDs

duplicate_ticket_ids = df["Ticket ID"].duplicated().sum()

print("Duplicate Ticket IDs:", duplicate_ticket_ids)
```

```python
# Compare ticket status with resolution time availability

print(
    df.groupby("Ticket Status")["Time to Resolution"].apply(
        lambda x: x.isnull().sum()
    )
)
```

```python
# Check data types

print(df.dtypes)
```

```python
# Find columns containing date or time

date_columns = [
    col for col in df.columns
    if "date" in col.lower() or "time" in col.lower()
]

print("Date and Time Columns:", date_columns)
```

```python
# Convert  Date of Purchase to datetime

df["Ticket Date"] = pd.to_datetime(
    df["Date of Purchase"],
    errors="coerce"
)

print(df["Date of Purchase"].dtype)
```

```python
# Convert date and time columns

date_columns = [
    "Date of Purchase",
    "First Response Time",
    "Time to Resolution"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

print("Date conversion completed!")
```

```python
print(df[date_columns].dtypes)
```

```python
df[date_columns].head(10)
```

```python
# Convert Ticket ID to string
df["Ticket ID"] = df["Ticket ID"].astype("string")

# Convert Customer Age to numeric integer
df["Customer Age"] = pd.to_numeric(
    df["Customer Age"],
    errors="coerce"
).astype("Int64")

# Convert Customer Satisfaction Rating
df["Customer Satisfaction Rating"] = pd.to_numeric(
    df["Customer Satisfaction Rating"],
    errors="coerce"
).astype("Int64")
```

```python
datatype_report = pd.DataFrame({
    "Column_Name": df.columns,
    "Data_Type": df.dtypes.astype(str).values
})

datatype_report
```

```python
# Step 10: Inconsistent Values Analysis
```

```python
# Check unique values in categorical columns

categorical_columns = [
    "Ticket Type",
    "Ticket Status",
    "Ticket Priority",
    "Ticket Channel",
    "Customer Gender"
]

for col in categorical_columns:

    print("\nColumn:", col)

    print(df[col].value_counts(dropna=False))
```

```python
# Check unique values after removing extra spaces

for col in categorical_columns:

    print("\nColumn:", col)

    print(
        df[col].astype("string").str.strip().unique()
    )
```

```python
# Check minimum and maximum customer age
print("Minimum Age:", df["Customer Age"].min())
print("Maximum Age:", df["Customer Age"].max())

# Check ages below 18 or above 100
invalid_ages = df[
    (df["Customer Age"] < 18) |
    (df["Customer Age"] > 100)
]

print("\nNumber of unusual age records:", len(invalid_ages))

# Display those records
invalid_ages[["Ticket ID", "Customer Age"]].head(10)
```

```python
# Check unique customer satisfaction ratings
print("Unique Ratings:")
print(sorted(df["Customer Satisfaction Rating"].dropna().unique()))

# Check minimum and maximum ratings
print("\nMinimum Rating:", df["Customer Satisfaction Rating"].min())
print("Maximum Rating:", df["Customer Satisfaction Rating"].max())

# Find ratings outside the expected 1–5 range
invalid_ratings = df[
    (df["Customer Satisfaction Rating"] < 1) |
    (df["Customer Satisfaction Rating"] > 5)
]

print("\nNumber of invalid ratings:", len(invalid_ratings))
```

```python
# Count descriptions containing the product placeholder
placeholder_count = df["Ticket Description"].str.contains(
    r"\{product_purchased\}",
    case=False,
    na=False
).sum()

print("Descriptions containing placeholder:", placeholder_count)

# Display some examples
df.loc[
    df["Ticket Description"].str.contains(
        r"\{product_purchased\}",
        case=False,
        na=False
    ),
    ["Product Purchased", "Ticket Description"]
].head(5)
```

```python
# Check unique values in categorical columns

categorical_columns = [
    "Ticket Type",
    "Ticket Status",
    "Ticket Priority",
    "Ticket Channel",
    "Customer Gender"
]

for col in categorical_columns:

    print("\nColumn:", col)

    print(df[col].value_counts(dropna=False))
```

```python
# Count repeated ticket descriptions
duplicate_descriptions = df["Ticket Description"].duplicated().sum()

print("Repeated ticket descriptions:", duplicate_descriptions)

# Display examples of repeated descriptions
df[df["Ticket Description"].duplicated(keep=False)][
    ["Ticket ID", "Product Purchased", "Ticket Description"]
].head(10)
```

```python
# Check missing values for each ticket status
missing_by_status = df.groupby("Ticket Status").agg(
    Total_Tickets=("Ticket ID", "count"),
    Missing_First_Response=("First Response Time", lambda x: x.isna().sum()),
    Missing_Resolution_Time=("Time to Resolution", lambda x: x.isna().sum()),
    Missing_Satisfaction_Rating=("Customer Satisfaction Rating", lambda x: x.isna().sum()),
    Missing_Resolution=("Resolution", lambda x: x.isna().sum())
)

missing_by_status
```

```python
# Check minimum and maximum dates
date_columns = [
    "Date of Purchase",
    "Ticket Date",
    "First Response Time",
    "Time to Resolution"
]

for col in date_columns:
    print("\nColumn:", col)
    print("Earliest:", df[col].min())
    print("Latest:", df[col].max())
```

```python
# Check whether response and resolution timestamps occur
# before or after the ticket date

print("First Response Before Ticket Date:")
print((df["First Response Time"] < df["Ticket Date"]).sum())

print("\nFirst Response After Ticket Date:")
print((df["First Response Time"] > df["Ticket Date"]).sum())

print("\nResolution Before Ticket Date:")
print((df["Time to Resolution"] < df["Ticket Date"]).sum())

print("\nResolution After Ticket Date:")
print((df["Time to Resolution"] > df["Ticket Date"]).sum())
```

```python
# Check unique values in categorical columns

categorical_columns = [
    "Ticket Type",
    "Ticket Status",
    "Ticket Priority",
    "Ticket Channel",
    "Customer Gender"
]

for col in categorical_columns:

    print("\nColumn:", col)

    print(df[col].value_counts(dropna=False))
```
